```python
import delta_sharing

# Credential file from the recipient activation link (endpoint + bearer token).
PROFILE = "config.share"

client = delta_sharing.SharingClient(PROFILE)

# Discover everything this recipient can read.
tables = client.list_all_tables()
if not tables:
    raise ValueError("No tables shared with this recipient.")

for t in tables:
    print(f"{t.share}.{t.schema}.{t.name}")

# Table address: <profile>#<share>.<schema>.<table>
def table_url(t):
    return f"{PROFILE}#{t.share}.{t.schema}.{t.name}"

# Load the first shared table (change index or loop to load others).
t = tables[0]
print(f"\nLoading: {t.share}.{t.schema}.{t.name}")
df = delta_sharing.load_as_pandas(table_url(t))
print(f"Rows: {len(df):,}  Columns: {len(df.columns)}")
df.head()
```
